```python
from coin import Coin
```

```python
from random import randint
```

```python
from collections import defaultdict
```

```python
# Defines a coin

coin = Coin("Heads", "Tails")
```

```python
# Main code

dd = defaultdict(int)
n_rounds = 100000

for _ in range(n_rounds):
    n_flips = randint(2,1000)
    for _ in range(n_flips):
        # checks the coin value
        coin.flip()
        
    # save the coin last value    
    dd[coin.value] += 1
```
